```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/bank.csv', sep=None, engine='python')

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python

print("Dataset Information:")
df.info()

print("\nMissing Values in Each Column:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nStatistical Summary:")
df.describe()
```

```python

print("Target Column Counts:")
print(df['y'].value_counts())

print("\nTarget Column Percentages:")
print(df['y'].value_counts(normalize=True).mul(100).round(2))

plt.figure(figsize=(7, 5))
sns.countplot(data=df, x='y')
plt.title("Bank Marketing Campaign - Target Distribution")
plt.xlabel("Subscribed to Term Deposit")
plt.ylabel("Number of Customers")
plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.histplot(data=df, x='age', bins=20, kde=True)

plt.title("Age Distribution of Bank Customers")
plt.xlabel("Age (Years)")
plt.ylabel("Number of Customers")

plt.show()
```

```python

job_counts = df['job'].value_counts()

print("Customer Count by Job:")
print(job_counts)

plt.figure(figsize=(10, 5))

sns.countplot(data=df, x='job', order=job_counts.index)

plt.title("Customer Distribution by Job")
plt.xlabel("Job Type")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

```python

marital_counts = df['marital'].value_counts()

print("Customer Count by Marital Status:")
print(marital_counts)

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='marital', order=marital_counts.index)

plt.title("Customer Distribution by Marital Status")
plt.xlabel("Marital Status")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.show()
```

```python

education_counts = df['education'].value_counts()

print("Customer Count by Education:")
print(education_counts)

plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='education', order=education_counts.index)

plt.title("Customer Distribution by Education")
plt.xlabel("Education Level")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.show()
```

```python

housing_counts = df['housing'].value_counts()

print("Housing Loan Distribution:")
print(housing_counts)

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='housing', order=housing_counts.index)

plt.title("Housing Loan Distribution")
plt.xlabel("Has Housing Loan?")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.show()
```

```python

loan_counts = df['loan'].value_counts()

print("Personal Loan Distribution:")
print(loan_counts)

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='loan', order=loan_counts.index)

plt.title("Personal Loan Distribution")
plt.xlabel("Has Personal Loan?")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.show()
```

```python

contact_counts = df['contact'].value_counts()

print("Customer Count by Contact Method:")
print(contact_counts)

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='contact', order=contact_counts.index)

plt.title("Customer Distribution by Contact Method")
plt.xlabel("Contact Method")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.show()
```

```python

poutcome_counts = df['poutcome'].value_counts()

print("Previous Campaign Outcome:")
print(poutcome_counts)

plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='poutcome', order=poutcome_counts.index)

plt.title("Previous Marketing Campaign Outcome")
plt.xlabel("Previous Campaign Outcome")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.show()
```

```python

# Subscription rate by previous campaign outcome

outcome_rate = pd.crosstab(
    df['poutcome'],
    df['y'],
    normalize='index'
) * 100

print("Subscription Rate by Previous Campaign Outcome:")
print(outcome_rate.round(2))

outcome_rate.plot(
    kind='bar',
    figsize=(9, 5),
    stacked=False
)

plt.title("Subscription Rate by Previous Campaign Outcome")
plt.xlabel("Previous Campaign Outcome")
plt.ylabel("Percentage of Customers")
plt.xticks(rotation=0)
plt.legend(title="Subscribed (y)")

plt.tight_layout()
plt.show()
```

## Final Conclusion – Bank Marketing EDA

The Bank Marketing dataset was analyzed using Python, Pandas, Matplotlib, and Seaborn. The analysis included dataset inspection, missing-value and duplicate checks, statistical summaries, and visualizations of customer characteristics.

### Key Findings

1. The dataset contains 4,521 customer records and 17 columns.
2. The target variable is imbalanced: 88.48% of customers did not subscribe to a term deposit, while 11.52% subscribed.
3. Most customers are between approximately 30 and 40 years old.
4. Management, blue-collar, and technician are the most common job categories.
5. Married customers form the largest marital-status group, and secondary education is the most common education level.
6. More customers have a housing loan than do not have one, while most customers do not have a personal loan.
7. Cellular is the most frequently recorded contact method.
8. Previous campaign success is associated with a substantially higher current subscription rate (64.34%) than failure (12.86%) or unknown outcome (9.10%).

### Conclusion

This exploratory data analysis identifies important patterns in customer demographics, financial characteristics, contact methods, and marketing outcomes. Previous campaign outcome appears particularly useful for understanding subscription behavior. These findings can help guide further analysis of customer response to bank marketing campaigns.

The analysis describes associations in the dataset and does not establish causation.
